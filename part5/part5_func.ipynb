### Warm-up: Small helpers

```python
# Area of circle
def area_of_circle(r):
    return 3.14 * r * r
print("Area of circle with radius 3:", area_of_circle(3))
print("Area of circle with radius 5:", area_of_circle(5))
print("Area of circle with radius 7:", area_of_circle(7))
```

```python
# Check if a number is even
def is_even(n):
    return n % 2 == 0
print("Is 4 even?", is_even(4))
print("Is 7 even?", is_even(7)) 
print("Is 10 even?", is_even(10))
```

```python
# Greet in different languages
def greet(name, lang="en"):
    if lang == "en":
        return f"Hello, {name}"
    elif lang == "ur":
        return f"Salaam, {name}"
    
print(greet("Ali"))
print(greet("Ali","ur"))
print(greet("Laiba","ur"))
```

### Build: Marks toolkit

```python
# Calculate average of marks
def average(marks):
    return sum(marks) / len(marks)
```

```python
def grade(avg: float) -> str:
    # Reject marks outside the valid range
    if avg < 0 or avg > 100:
        raise ValueError("Marks must be between 0 and 100.")

    if avg >= 80:
        return "A"
    elif avg >= 70:
        return "B"
    elif avg >= 60:
        return "C"
    elif avg >= 33:
        return "D"
    else:
        return "Fail"
```

```python
# Report student average and grade
def report(name, marks):
    avg = average(marks)
    student_grade = grade(avg)

    return f"{name}: Average = {avg:.2f}, Grade = {student_grade}"
```

```python
# Three students
print(report("Ali", [85, 90, 80]))
print(report("Sara", [70, 75, 80]))
print(report("Ahmed", [45, 55, 60]))
```
