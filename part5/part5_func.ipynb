### Warm-up: Small helpers

```python
# Area of circle
def area_of_circle(r):
    return 3.14 * r * r
print("Area of circle with radius 3:", area_of_circle(3))
print("Area of circle with radius 5:", area_of_circle(5))
print("Area of circle with radius 7:", area_of_circle(7))
```

```python
# Check if a number is even
def is_even(n):
    return n % 2 == 0
print("Is 4 even?", is_even(4))
print("Is 7 even?", is_even(7)) 
print("Is 10 even?", is_even(10))
```

```python
def greet(name, lang="en"):
    if lang == "en":
        return f"Hello, {name}"
    elif lang == "ur":
        return f"Salaam, {name}"
    
greet("Ali","en")
greet("Ali","ur")
greet("Laiba","ur")
```
