### Warm-up: Countdown generato

```python
# Generator that counts down from n to 1
def countdown(n):
    for i in range(n, 0, -1):
        yield i


# Display countdown
print("--- Countdown ---")

for number in countdown(5):
    print(number)


# Generator that produces even numbers
def evens(limit):
    for i in range(0, limit + 1, 2):
        yield i


# Display even numbers as a list
print("\n--- Even Numbers ---")
print(list(evens(10)))
```

### Build : Big file reader

```python

# Name: Kashaf
# Part 10 - Build: Big File Reader

# Step 1: Create a text file with 50 lines
with open("logs.txt", "w") as file:
    for i in range(1, 51):
        if i % 5 == 0:
            file.write(f"Line {i}: ERROR - Something went wrong\n")
        else:
            file.write(f"Line {i}: System running normally\n")


# Step 2: Generator to read the file one line at a time
def read_lines(path):
    with open(path, "r") as file:
        for line in file:
            yield line


# Step 3: Generator to produce only ERROR lines
def only_errors(lines):
    for line in lines:
        if "ERROR" in line:
            yield line


# Step 4: Count ERROR lines without creating a list
error_count = 0

for line in only_errors(read_lines("logs.txt")):
    print(line.strip())
    error_count += 1


# Step 5: Display total ERROR lines
print("\nTotal ERROR lines:", error_count)
```
