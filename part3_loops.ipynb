# 3 · Loops (for / while)

### Warm-up: Times table

```python
number = int(input("Enter a number: "))

for i in range (1,11):
    result = number * i
    print(f"{number} x {i} = {result}")
```

### Build: ATM PIN system

```python
for i in range (1,4):
    atm_pin= int(input("Enter your ATM pin: "))
    if atm_pin == 4321:
        print ("Welcome!")
        break
    else:
        print ("Incorrect PIN")
        print ("Attempts remaining:", 3-i)

else:
    print("Card blocked")
```
