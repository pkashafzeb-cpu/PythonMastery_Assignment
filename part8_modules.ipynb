### Warm-up : Standard library tour

```python
import math
import random
import datetime
import statistics
```

```python
# Math module
print ("square root of 144:", math.sqrt(144))
```

```python
# Random module
students = ["Ali", "Sara","Saba","Ahmad","Laiba"]
print("Random student name from the list:", random.choice(students))
```

```python
# Datetime module

# today's date
today = datetime.date.today()

# end of the year date
end_date = datetime.date(today.year, 12, 31)

# calculate the number of days left until December 31
days_left = (end_date - today).days
print("Days left until December 31:", days_left)
```

```python
# Statistics module
marks = [85, 90, 75, 80, 95]

print("Mean:", statistics.mean(marks))
print("Median:", statistics.median(marks))
```
