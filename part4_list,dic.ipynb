```python
contacts = {
    "Ali": "03001234567",
    "Sara": "03111234567",
    "Ahmed": "03221234567",
    "Ayesha": "03331234567",
    "Usman": "03441234567"
}

# Add a new contact
name = input("Enter contact name: ")
phone = input("Enter phone number: ")

contacts[name] = phone

print(contacts)
print("Contact added successfully!")

```

```python
# Search for a contact
name_search = input ("Enter name to search:")

print(contacts.get(name_search,"Contact not found"))
```

```python
# Delete a contact
name_del = input ("Enter name to delete: ")

contacts.pop(name_del,None)

print(contacts)

```

```python
# Show all contacts
for name, phone in contacts.items():
    print (f"Name: {name}, Phone: {phone}")
```

```python

contacts = {
    "Ali": "03001234567",
    "Sara": "03111234567",
    "Ahmed": "03221234567",
    "Ayesha": "03331234567",
    "Usman": "03441234567"
}

while True:
    print("\n--- Phone Book ---")
    print("1. Add contact")
    print("2. Search contact")
    print("3. Delete contact")
    print("4. Show all contacts")
    print("5. Exit")

    choice = input("Enter your choice: ")


    if choice == "1":
        name = input("Enter contact name: ")
        phone = input("Enter phone number: ")

        contacts[name] = phone
        print("Contact added successfully!")

    elif choice == "2":
        name_search = input("Enter name to search: ")
        print(contacts.get(name_search, "Not found"))

    elif choice == "3":
        name_del = input("Enter name to delete: ")

        if name_del in contacts:
            contacts.pop(name_del)
            print("Contact deleted successfully!")
        else:
            print("Contact not found!")

    elif choice == "4":
        for name, phone in contacts.items():
            print(f"Name: {name}, Phone: {phone}")

    elif choice == "5":
        print("Goodbye!")
        break

    else:
        print("Invalid choice. Try again.")
```
