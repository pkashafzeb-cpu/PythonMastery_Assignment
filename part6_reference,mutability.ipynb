### Build: Fix the cart bug

```python
def add_item(item, cart=[]):
    cart.append(item)
    return cart

print(add_item("Milk"))
print(add_item("Bread"))
```

```python
# Fix the cart bug
def add_item_fixed(item, cart=None):
    if cart is None:
        cart = []

    cart.append(item)
    return cart

print(add_item_fixed("Milk"))
print(add_item_fixed("Bread"))
```
