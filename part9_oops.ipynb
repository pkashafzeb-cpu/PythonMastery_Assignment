### Warm-up: Student class

```python
class Student:
    def __init__(self, name, roll, marks):
        
        self.name = name
        self.roll = roll
        self.marks = marks

# ADD marks method
    def add_marks(self, new_marks):
        self.marks.append(new_marks)

# average method
    def average(self):
     return sum(self.marks) / len(self.marks)

# str method
    def __str__(self):
     return f"{self.name} (Roll {self.roll}): avg {self.average():.1f}"

# Create student object
student1 = Student("Ali", 12, [80, 90, 70])

# Add a new mark
student1.add_marks(85)

# Display student information
print(student1)
print("Average:", student1.average())
print("Marks:", student1.marks)
```

### Build: Bank Account.

```python

class BankAccount:

    # Initialize account owner, balance, and transaction history
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance
        self.history = []

    # Deposit money into the account
    def deposit(self, amount):

        # Reject zero or negative amounts
        if amount <= 0:
            raise ValueError("Deposit amount must be positive.")

        # Update balance
        self.balance += amount

        # Record transaction
        self.history.append(f"Deposited: Rs. {amount}")

        return self.balance

    # Withdraw money from the account
    def withdraw(self, amount):

        # Reject zero or negative amounts
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")

        # Reject withdrawal greater than available balance
        if amount > self.balance:
            raise ValueError("Insufficient balance.")

        # Update balance
        self.balance -= amount

        # Record transaction
        self.history.append(f"Withdrawn: Rs. {amount}")

        return self.balance

    # Display transaction history
    def show_history(self):
        print("\n--- Transaction History ---")

        for transaction in self.history:
            print(transaction)


# Create a bank account object
account1 = BankAccount("Ali", 5000)

# Deposit money
account1.deposit(2000)

# Withdraw money
account1.withdraw(1500)

# Display account information
print("Owner:", account1.owner)
print("Current Balance: Rs.", account1.balance)

# Display transaction history
account1.show_history()
```
