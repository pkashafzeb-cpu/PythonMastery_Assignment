### Student ID card:

### Build Shop bill

```python
item = str(input("Enter the name of the item: "))
price = float(input("Enter the price of the item: "))
quantity = int(input("Enter the quantity of the item: "))
```

```python
#SUBTOTAL
subtotal= price * quantity

#5% discount
discount = subtotal * 0.05

# Amount after discount
discounted_amount = subtotal - discount

# tax amount
tax = discounted_amount * 0.17

#final total
final_bill = subtotal - discount + tax
```

```python
print ("\n--- Shop Bill ---")
print("Item:", item)
print("Price:", round(price, 2))
print("Quantity:", quantity)
print("Sub-total:", round(subtotal, 2))
print("5% Discount:", round(discount, 2))
print("Amount after Discount:", round(discounted_amount, 2))
print("Tax:", round(tax, 2))
print("Final Bill:", round(final_bill, 2))
```

### Challenge — Unit converter

```python
temp = float(input("Enter the temperature in Celsius: "))
# Convert Celsius to Fahrenheit
Fahrenheit = (temp * 9/5) + 32

#Convert Celsius to Kelvin
Kelvin = temp + 273.15
print (f"Temperature in Celsius:{temp:.1f}")
print (f"Temperature in Fahrenheit:{Fahrenheit:.1f}")
print (f"Temperature in Kelvin:{Kelvin:.1f}")
```

```python
# Distance Conversion
dist_km = float(input("Enter the distance in kilometers: "))

# Convert kilometers to miles
dist_miles = dist_km * 0.621371

# Convert kilometers to meters
dist_meters = dist_km * 1000

print(f"Distance in kilometers: {dist_km :.1f}")
print(f"Distance in miles: {dist_miles:.1f}")
print(f"Distance in meters: {dist_meters:.1f}")
```
